# Generate topobathy GeoTIFF from GEBCO

Download a regional subset from [GEBCO 2025](https://www.gebco.net/data-products/gridded-bathymetry-data/gebco2025-grid) (CEDA OPeNDAP) at **native grid resolution** (~400 m), save a GeoTIFF.

Use this notebook when you do not have local bathymetry yet.

Downstream use:

1. Point `paths.aoi_file` and `paths.topobathy_file` in the case JSON to the correspondent paths.
2. Continue with `02_Create_mesh_by_bluemesh2d.ipynb`.

Requires network access for the GEBCO download.

````python
import geopandas as gpd
import matplotlib.pyplot as plt
import rioxarray
import xarray as xr
from shapely.geometry import Polygon
from utils.config import PATHS, case_name, resolve
````

## Parameters

Two AoI modes are supported:

1. **Use existing AoI shapefile** from `paths.aoi_file`
2. **Create AoI shapefile from vertices** (entered clockwise)

In both cases, GEBCO bounds are derived automatically as AoI bounds + 50 km buffer.

````python
# Padding around AoI bounds for GEBCO subset
buffer_km = 50.0

# AoI mode:
# - True  -> read existing shapefile from paths.aoi_file
# - False -> create shapefile from vertices_clockwise below
use_existing_aoi = True

# Required only when use_existing_aoi = False.
# Enter lon/lat vertices in clockwise order and close the polygon by repeating
# the first vertex at the end (left-to-right progression around the AoI).
vertices_clockwise = [
    # (-171.95, -13.78),
    # (-171.55, -13.78),
    # (-171.55, -14.15),
    # (-171.95, -14.15),
    # (-171.95, -13.78),
]

# GEBCO 2025 OPeNDAP endpoint (ice-surface / land elevation)
gebco_url = (
    "https://dap.ceda.ac.uk/thredds/dodsC/bodc/gebco/global/gebco_2025/"
    "ice_surface_elevation/netcdf/GEBCO_2025.nc"
)
````

````python
# Inputs/outputs
input_aoi = resolve(PATHS["aoi_file"])
output_tiff = resolve(f"inputs/GEBCO/{case_name()}_GEBCO.tiff")

output_tiff.parent.mkdir(parents=True, exist_ok=True)

print(f"AoI shp  : {input_aoi}")
print(f"GeoTIFF  : {output_tiff}")
print(f"buffer km: {buffer_km}")
````

## Build/Load AoI and derive buffered bounds

- If `use_existing_aoi=True`, read `paths.aoi_file`.
- If `use_existing_aoi=False`, create `paths.aoi_file` from `vertices_clockwise`.

Then expand outer bounds by 50 km (in projected meters) before subsetting GEBCO.

````python
if use_existing_aoi:
    if not input_aoi.exists():
        raise FileNotFoundError(
            f"AoI file not found: {input_aoi}\n"
            "Set use_existing_aoi=False to create it from vertices_clockwise."
        )
    aoi_gdf = gpd.read_file(input_aoi).to_crs("EPSG:4326")
    print(f"Loaded existing AoI: {input_aoi}")
else:
    if len(vertices_clockwise) < 4:
        raise ValueError(
            "vertices_clockwise must contain at least 4 points (including closure)."
        )

    polygon = Polygon(vertices_clockwise)
    if not polygon.is_valid:
        raise ValueError("Provided vertices_clockwise do not form a valid polygon.")

    if not polygon.exterior.is_ccw:
        # As requested: vertices should be clockwise.
        pass
    else:
        print("Warning: vertices are counter-clockwise; proceeding anyway.")

    input_aoi.parent.mkdir(parents=True, exist_ok=True)
    aoi_gdf = gpd.GeoDataFrame({"name": ["aoi"]}, geometry=[polygon], crs="EPSG:4326")
    aoi_gdf.to_file(input_aoi)
    print(f"Created AoI shapefile: {input_aoi}")

# Buffer in meters using Web Mercator, then return to lon/lat.
buffered = aoi_gdf.to_crs("EPSG:3857").buffer(buffer_km * 1000.0).to_crs("EPSG:4326")

bbox = dict(zip(["lon_min", "lat_min", "lon_max", "lat_max"], buffered.total_bounds))
print({k: round(v, 5) for k, v in bbox.items()})
aoi_gdf
````

## Download and subset GEBCO

````python
GEBCO = xr.open_dataset("inputs/GEBCO/bathy_nsw_blended_50m.nc")
GEBCO
````

````python
elev_name = "elevation" if "elevation" in GEBCO else list(GEBCO.data_vars)[0]
print(f"elevation variable: {elev_name}")

# GEBCO latitude is usually descending — slice high to low
lat_slice = slice(bbox["lat_min"], bbox["lat_max"])
lon_slice = slice(bbox["lon_min"], bbox["lon_max"])

gebco_sel = GEBCO.sel(lon=lon_slice, lat=lat_slice)
gebco_sel
````

## Write GeoTIFF (native GEBCO resolution)

````python
gebco_raster = (
    -GEBCO["depth"]
    .sortby("lat", ascending=False)
    .rio.set_spatial_dims(x_dim="lon", y_dim="lat", inplace=False)
    .rio.write_crs("EPSG:4326", inplace=False)
)

gebco_raster.rio.to_raster(output_tiff)
print(f"Wrote {output_tiff}")
print(f"CRS: {gebco_raster.rio.crs}")
print(f"Resolution (deg): {gebco_raster.rio.resolution()}")
print(f"Shape: {gebco_raster.shape}")
````

## Quick preview

````python
topobathy = rioxarray.open_rasterio(output_tiff).squeeze(drop=True)

fig, ax = plt.subplots(figsize=(9, 6))
topobathy.plot.imshow(ax=ax, vmin=-200, vmax=0, cmap="terrain")
aoi_gdf.boundary.plot(ax=ax, color="black", linewidth=1.2)
ax.set_title(output_tiff.name)
fig.tight_layout()
plt.show()
````

## Optional — paste a higher-resolution DEM on top of GEBCO

When a better local DEM exists (e.g. the NOAA coastal DEMs on
[NCEI THREDDS](https://www.ngdc.noaa.gov/thredds/catalog/regional/catalog.html)),
build a single merged GeoTIFF:

- the **target grid is the DEM grid** (native resolution), extended outward at
  the same step to cover the union of the GEBCO tiff extent and the DEM extent —
  so the good data is written **untouched**, no resampling;
- GEBCO (already open above) is bilinearly upsampled to fill everywhere else.

Notes:

- The paste is a hard seam; datasets can disagree in deep water (harmless for
  wave modelling — nearshore is where the DEM wins).
- Vertical datums may differ (e.g. NOAA coastal DEMs are MHW, GEBCO ~MSL).
- To use the merged file downstream, point `paths.topobathy_file` in the case
  JSON at the merged GeoTIFF.

Skip this section if GEBCO alone is enough for your case.


````python
# --- merged-DEM parameters ---
# Higher-resolution local DEM (OPeNDAP URL or local NetCDF path) and its variable.
# Example: a NOAA coastal DEM from https://www.ngdc.noaa.gov/thredds (var "Band1").
dem_url = "TODO_DEM_URL"
dem_var = "Band1"
merged_tiff = resolve(f"inputs/GEBCO/{case_name()}_GEBCO_merged.tiff")

print(f"DEM    : {dem_url}")
print(f"merged : {merged_tiff}")
````

````python
import numpy as np

# Local DEM (lat ascending) and its grid step
dem = xr.open_dataset(dem_url)[dem_var].load().sortby("lat")
step = float(np.round(np.diff(dem.lon.values).mean(), 12))

# Union bbox: GEBCO tiff extent (generated above) + DEM extent
gebco_tiff = rioxarray.open_rasterio(output_tiff)
gw, gs, ge, gn = gebco_tiff.rio.bounds()
west, south = min(gw, float(dem.lon.min())), min(gs, float(dem.lat.min()))
east, north = max(ge, float(dem.lon.max())), max(gn, float(dem.lat.max()))


def extend(coords, lo, hi):
    """Extend DEM coords outward at the same step until [lo, hi] is covered."""
    n_lo = max(0, int(np.ceil((coords[0] - lo) / step)))
    n_hi = max(0, int(np.ceil((hi - coords[-1]) / step)))
    return np.concatenate(
        [
            coords[0] - step * np.arange(n_lo, 0, -1),
            coords,
            coords[-1] + step * np.arange(1, n_hi + 1),
        ]
    )


target_lon = extend(dem.lon.values, west, east)
target_lat = extend(dem.lat.values, south, north)
print(f"DEM grid step : {step:.9f} deg (~{step * 111320:.0f} m)")
print(f"bbox          : lon [{west:.4f}, {east:.4f}], lat [{south:.4f}, {north:.4f}]")
print(f"target grid   : {len(target_lat)} x {len(target_lon)}")
````

````python
# GEBCO subset (small pad for edge support), upsampled to the target grid
pad = 0.02
gebco_sub = (
    GEBCO[elev_name]
    .sel(lon=slice(west - pad, east + pad), lat=slice(south - pad, north + pad))
    .load()
    .sortby("lat")
)
gebco_hi = gebco_sub.interp(
    lon=target_lon, lat=target_lat, method="linear", kwargs={"fill_value": None}
).astype("float32")

# Paste the DEM on top; GEBCO fills the rest
dem_on_target = dem.reindex(
    lon=target_lon, lat=target_lat, method="nearest", tolerance=step / 2
)
merged = dem_on_target.combine_first(gebco_hi)

n_dem = int(dem_on_target.notnull().sum())
print(f"cells from DEM: {n_dem} / {merged.size} ({100 * n_dem / merged.size:.1f}%)")
assert int(merged.isnull().sum()) == 0, "merged raster contains NaNs"
````

````python
merged = (
    merged.sortby("lat", ascending=False)
    .rio.set_spatial_dims(x_dim="lon", y_dim="lat")
    .rio.write_crs("EPSG:4326")
)
merged.rio.to_raster(merged_tiff)
print(f"Wrote {merged_tiff}")
print(f"Resolution (deg): {merged.rio.resolution()}")
print(f"Shape: {merged.shape}")
````

````python
merged_prev = rioxarray.open_rasterio(merged_tiff).squeeze(drop=True)

fig, ax = plt.subplots(figsize=(10, 6))
merged_prev.plot.imshow(ax=ax, vmin=-6000, vmax=1000, cmap="terrain")
aoi_gdf.boundary.plot(ax=ax, color="black", linewidth=1.2)
ax.set_title(merged_tiff.name)
fig.tight_layout()
plt.show()
````

If you built the optional merged DEM above, point `topobathy_file` at the merged GeoTIFF instead.
