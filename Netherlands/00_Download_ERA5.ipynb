```python
import xarray as xr

PAT = "edh_pat_6e44645d42ae5997121872e8416920a6c3ab0bb4da924ad6457b9741efde7d2d5cea9d1581f03c49c1fd4ed6381daa8e"

era5 = xr.open_dataset(
    f"https://edh:{PAT}@data.earthdatahub.destine.eu/era5/reanalysis-era5-single-levels-ocean-v0.zarr",
    storage_options={"client_kwargs": {"trust_env": True}},
    chunks={},
    engine="zarr",
)
era5
```

```python
import numpy as np
import xarray as xr

era5_dano = xr.open_dataset("inputs/ERA5_semiRawData_2000_2025.nc")
era5_dano
```

```python
era5_dano_locations = (
    era5_dano[
        [
            "wmb",
            "swh",
            "mwp",
            "pp1d",
            "mwd",
            "wdw",
            "shww",
            "mpww",
            "mdww",
            "dwww",
            "swh1",
            "mwp1",
            "mwd1",
            "dwps",
            "swh2",
            "mwp2",
            "mwd2",
            "swh3",
            "mwp3",
            "mwd3",
        ]
    ]
    .stack(location=("latitude", "longitude"))
    .dropna(dim="location")
    .isel(location=[46, 43, 40, 35, 26, 16, 17, 18, 19, 20][::-1])
    .rename(
        {
            "valid_time": "time",
            "location": "goal",
        }
    )
    .drop(["meanSea", "number"])
)
era5_dano_locations_latitude = era5_dano_locations.latitude.values
era5_dano_locations_longitude = era5_dano_locations.longitude.values
era5_dano_locations = era5_dano_locations.drop(["latitude", "longitude"])
era5_dano_locations["goal"] = range(1, 11)
era5_dano_locations = era5_dano_locations.assign_coords(
    {
        "latitude": (("goal"), era5_dano_locations_latitude),
        "longitude": (("goal"), era5_dano_locations_longitude),
    }
)

era5_dano_locations["wdw"] = era5_dano_locations["wdw"] * 180 / np.pi
era5_dano_locations["dwww"] = era5_dano_locations["dwww"] * 180 / np.pi
era5_dano_locations["dwps"] = era5_dano_locations["dwps"] * 180 / np.pi

era5_dano_locations
```

```python
era5_dano_locations.to_netcdf("inputs/ERA5_forcing_goals_2000_2025.nc")
```

```python
era5_dano_locations.isel(goal=1)["swh1"].plot()
era5_dano_locations.isel(goal=1)["swh2"].plot()
era5_dano_locations.isel(goal=1)["swh3"].plot()
```
